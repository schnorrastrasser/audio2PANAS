Damit die Daten nützlich sind, beachte folgende strikte Vorgaben:

1. PSYCHOLOGISCHE KONSISTENZ BEI MULTIDIMENSIONALITÄT: Da die Texte komplexe, multidimensionale Gefühlslagen abbilden, müssen die Scores diese Ambivalenz exakt widerspiegeln. Jemand kann z. Setze unausgesprochene Facetten dann NICHT einfach pauschal auf 1 oder 3. Erzeuge eine realistische, menschliche Varianz, die zur Grundstimmung passt. 3. ZUSATZ-METRIKEN: Berechne ganz am Ende der Spalten zusätzlich "PA_Mean" (Durchschnitt der 10 Positiver-Affekt-Items) und "NA_Mean" (Durchschnitt der 10 Negativer-Affekt-Items) pro Text.
4. VERARBEITUNG IN SCHRITTEN: Verarbeite bitte zunächst NUR DIE ERSTEN 25 EINTRÄGE. Sobald ich "Weiter" schreibe, analysierst du nahtlos die nächsten 25 (26-50), usw.
5. FORMAT: Gib das Ergebnis AUSSCHLIESSLICH als gültigen CSV-Code im Code-Block aus (Trennzeichen: Semikolon). Nimm den exakten "Antwort_Text" aus der vorherigen Generierung unverändert in die Tabelle auf und setze ihn in doppelte Anführungszeichen.

Die CSV-Struktur MUSS exakt so aussehen (ohne Zeilenumbrüche innerhalb der Header-Zeile):
ID;Antwort_Text;aktiv;bekuemmert;interessiert;freudig_erregt;veraergert;stark;schuldig;erschrocken;feindselig;angeregt;stolz;gereizt;begeistert;beschaemt;wach;nervoes;entschlossen;aufmerksam;durcheinander;aengstlich;PA_Mean;NA_Mean

Bitte bestätige kurz, dass du die Reihenfolge verstanden hast, und generiere dann direkt die CSV-Tabelle für die IDs 1 bis 25. ich will 2 verschiede LLMs prompten dass sie mir dasselbe Ergebnis beim raten von den texten auch für die schätzung der unsicheren werte geben, wenn nicht sicher dann gerne auf 0 statt mit variant 

```python
import pandas as pd

data_a = pd.read_csv('/Users/annafelleiter/Documents/Uni/NLE2/EMA2AI/data/Synt_interrater_rel.csv', sep=';')
data_b = pd.read_csv('/Users/annafelleiter/Documents/Uni/NLE2/EMA2AI/data/Synt_fin.csv', sep=';')

data_a = data_a.drop(columns=['Inferenz_Notiz'], errors='ignore')

data_b = data_b[data_b['ID'] <= 350]

len(data_a), len(data_b)
```

```python
data_a.head()
```

```python
data_b.head()
```

```python
from sklearn.metrics import cohen_kappa_score

# 1. Alle Spalten ausschließen, die KEINE diskreten Rater-Kategorien sind
exclude_cols = ["ID", "Antwort_Text", "PA_Mean", "NA_Mean"]

results = []

for col in data_a.columns:
    # Optional: Verhindert Fehler, falls unbenannte Zusatzspalten (wie Anmerkungen) vorkommen
    if col.startswith("Unnamed") or col.startswith("Explizit"):
        continue

    if col not in exclude_cols and col in data_b.columns:
        # Paaren und NaN-Werte entfernen
        df_paired = pd.DataFrame(
            {"r1": data_a[col].values, "r2": data_b[col].values}
        ).dropna()

        # Werte explizit in Integers wandeln (behebt den ValueError)
        try:
            rater1 = df_paired["r1"].astype(int)
            rater2 = df_paired["r2"].astype(int)

            # Kappa berechnen
            unweighted = cohen_kappa_score(rater1, rater2)
            weighted = cohen_kappa_score(rater1, rater2, weights="quadratic")

            results.append(
                {
                    "Spalte": col,
                    "Unweighted_Kappa": round(unweighted, 4),
                    "Weighted_Kappa": round(weighted, 4),
                }
            )
        except ValueError:
            # Falls eine Spalte Text/Strings enthält, die sich nicht in int wandeln lassen
            print(f"Spalte '{col}' übersprungen (keine gültigen Zahlen).")

# Ergebnis als DataFrame
results_df = pd.DataFrame(results)
print(results_df)
```
